 ->>>>>>>>>1.Is there a statistically significant linear correlation between high temperatures and

so first we dont have the column name climate zone so we have to create it 

```python
def climatezone(lat):
    if abs(lat)<23.5:
        return "Tropical"
    elif abs(lat) < 40 :
        return "Subtropical"
    else:
        return "Temperate"



df['climate_zone']=df['latitude'].apply(climatezone)
```

```python
df['climate_zone'].isnull().sum()
```

```python
from scipy.stats import pearsonr

for zone in df['climate_zone'].unique():
    subset = df[df['climate_zone'] == zone]
    
    r, p = pearsonr(subset['temperature_celsius'], 
                    subset['humidity'])
    
    print(f"\nClimate Zone: {zone}")
    print("Correlation (r):", round(r, 3))
    print("P-value:", round(p, 5))
```

->>>>>>>>>>>>>>>>>>.18. After a peak "extreme weather event" (like a storm or heatwave), what is the average
"recovery time" for a region to return to its 30-day rolling mean temperature?

```python
df['rolling_mean_30'] = df.groupby('location_name')['temperature_celsius'].transform(lambda x: x.rolling(30).mean())


mean_temp = df['temperature_celsius'].mean()
std_temp = df['temperature_celsius'].std()

df['extreme_event'] = (
    abs(df['temperature_celsius'] - mean_temp) > 3 * std_temp
)


df['recovered'] = abs(df['temperature_celsius'] - df['rolling_mean_30']) < 1


df

extreme_days = df[df['extreme_event']]


extreme_days

recovery_times = []

for idx in extreme_days.index:
    
    location = df.loc[idx,'location_name']
    event_date = df.loc[idx,'last_updated']
    
    future = df[(df['location_name']==location) &
                (df['last_updated'] > event_date)]
    
    recovered_day = future[future['recovered']].head(1)
    
    if not recovered_day.empty:
        recovery = (recovered_day['last_updated'].values[0] - event_date).days
        recovery_times.append(recovery)


```
