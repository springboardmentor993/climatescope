```python
import pandas as pd
import matplotlib.pyplot as plt

df["last_updated"] = pd.to_datetime(df["last_updated"])
```

```python
df["Month"] = df["last_updated"].dt.month
```

```python
df[["last_updated", "Month"]].head()
```

```python
def get_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Spring"
    elif month in [6, 7, 8]:
        return "Summer"
    else:
        return "Autumn"

df["Season"] = df["Month"].apply(get_season)
```

```python
df[["Month", "Season"]].head()
```

```python
df["Season"].value_counts().plot(kind="bar")
plt.title("Season Distribution")
plt.xlabel("Season")
plt.ylabel("Count")
plt.show()
```

```python
avg_temp = df.groupby("country")["temperature_celsius"].mean().sort_values(ascending=False).head(10)

avg_temp.plot(kind="bar")
plt.title("Top 10 Countries with Highest Average Temperature")
plt.xlabel("Country")
plt.ylabel("Average Temperature (°C)")
plt.xticks(rotation=45)
plt.show()
```

```python
df["temperature_celsius"].plot(kind="hist", bins=30)
plt.title("Temperature Distribution")
plt.xlabel("Temperature (°C)")
plt.ylabel("Frequency")
plt.show()
```

```python
season_temp = df.groupby("Season")["temperature_celsius"].mean()

season_temp.plot(kind="bar")
plt.title("Average Temperature per Season")
plt.xlabel("Season")
plt.ylabel("Average Temperature (°C)")
plt.show()
```

```python
humid_places = df.groupby("location_name")["humidity"].mean().sort_values(ascending=False).head(10)

humid_places.plot(kind="bar")
plt.title("Top 10 Most Humid Locations")
plt.xlabel("Location")
plt.ylabel("Average Humidity")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure()
plt.scatter(df["temperature_celsius"], df["humidity"])
plt.title("Temperature vs Humidity")
plt.xlabel("Temperature (°C)")
plt.ylabel("Humidity (%)")
plt.show()
```

```python
aqi_avg = df.groupby("country")["air_quality_us-epa-index"].mean().sort_values(ascending=False).head(10)

plt.figure()
aqi_avg.plot(kind="bar")
plt.title("Top 10 Countries by Average US EPA AQI")
plt.xlabel("Country")
plt.ylabel("Average AQI")
plt.xticks(rotation=45)
plt.show()
```

```python
pivot = pd.crosstab(df["country"], df["Season"])

plt.figure()
plt.imshow(pivot)
plt.title("Country-wise Season Distribution Heatmap")
plt.xlabel("Season")
plt.ylabel("Country")
plt.colorbar()
plt.show()
```

```python
plt.figure()
correlation = df.corr(numeric_only=True)
plt.imshow(correlation)
plt.title("Correlation Matrix")
plt.colorbar()
plt.show()
```

```python
df["last_updated"] = pd.to_datetime(df["last_updated"])
df_sorted = df.sort_values("last_updated")

plt.figure()
df_sorted.groupby(df_sorted["last_updated"].dt.date)["temperature_celsius"].mean().plot()
plt.title("Temperature Trend Over Time")
plt.xlabel("Date")
plt.ylabel("Average Temperature (°C)")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure()
plt.scatter(df["temperature_celsius"], df["humidity"])
plt.title("Temperature vs Humidity")
plt.xlabel("Temperature (°C)")
plt.ylabel("Humidity (%)")
plt.show()
```

```python
plt.figure()
plt.scatter(df["temperature_celsius"], df["air_quality_us-epa-index"])
plt.title("Temperature vs Air Quality Index")
plt.xlabel("Temperature (°C)")
plt.ylabel("US EPA AQI")
plt.show()
```

```python
polluted = df.groupby("country")["air_quality_us-epa-index"].mean().sort_values(ascending=False).head(10)

plt.figure()
polluted.plot(kind="bar")
plt.title("Top 10 Most Polluted Countries")
plt.xlabel("Country")
plt.ylabel("Average AQI")
plt.xticks(rotation=45)
plt.show()
```
